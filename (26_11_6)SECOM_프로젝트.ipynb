.ipynb 파일은 Python 코드뿐 아니라 실행 결과, 그래프, 설명 문장까지 함께 저장할 수 있는 노트북 파일

```python

import pandas as pd

# 1. 가상의 반도체 공정 데이터 생성
data = {
    "sensor_001": [10, 12, 9, 11],
    "result": ["PASS", "PASS", "FAIL", "PASS"]
}

# 2. 표 형태로 변환
df = pd.DataFrame(data)
###pd.DataFrame()	데이터를 표로 만들기

# 3. 전체 데이터 확인
print("=== 반도체 공정 데이터 ===")
print(df)

# 4. 센서 평균 계산
print("\n센서 평균:", df["sensor_001"].mean())
###df["sensor_001"]	특정 센서 데이터 선택하기

# 5. 불량률 계산
fail_rate = (df["result"] == "FAIL").mean() * 100
###.mean()	평균 계산하기
###== "FAIL"	불량에 해당하는 데이터 찾기

print("불량률:", fail_rate, "%")
```
